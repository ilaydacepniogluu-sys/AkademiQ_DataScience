#Standardizasyon (daha sonraki adımlarda hata yapmamak için veriyi standartlaştır)


```python
dt.columns = (dt.columns
              .str.strip() #örn:  " Age " =  "Age" (metinsel ifadenin başındaki ve sonundaki boşlukları temizledi.)
              .str.lower() #CustomerID= customerid (tüm harfleri küçük hale çevirir.)
              .str.replace(" "," _") #customer name = customer_name
              .str.replace("-","_")
              )
print("\nTemizlenmiş sütun isimleri:")
#print(dt.columns.tolist())
print("\n".join(dt.columns)) #tüm sütunları tek satırda alt alta yazar yani her listede yazılan elemanın arasına /n ekler.
```

```python
string_cols= dt.select_dtypes(include="object").columns #string olan sütunlardan sıyrılmaya çalışıyoruz çünkü sayısal değerlerle işlem yapacağız.
#dtype ı obje olanların tamamını sütunlardan seç(üstteki kod açıklaması)
for x in string_cols: #her string sütununu tek tek dolaş (kod acıklaması)
    dt[x] = dt[x].astype(str).str.strip()
                 #Astype(str) bu pythonun kendi veri tipidir. Yani veri tipini dönüştürmek demek. Sütundaki tüm değerleri string hale getir demektir.
    # (strip)Obje türüne ait sütunlarda görünmeyen bir boşluk varsa eğer onu düzeltir. Baştaki ve sondaki boşlukları temizledi.
                            #.str ifadesi #123 gibi değer varsa onu "123" e dönüştürür. Ek olarak sadece string olan sütunlarda strip işlemini uygulamak için yazıldı.
                            #none = "none" (none string bir veri değildir. O veri yok anlamına gelir.)
#NOT: görünmeyen boşluklar veri analizinin en kritik hatalarından biridir.
```

```python
critical_cols = ["repository __name","stars __count"]
dt=dt.dropna(subset=critical_cols)
# Kritik sütun belirlemek demek bu sütun bizim için olmazsa olmaz demektir. Hiçbir şekilde kaldıralamaz drop yapılamaz.
#(subset=critical_cols) yukardaki sütundaki veriler girilmemiş eksik veri içeriyosa o satırı sil.
#dropna= drop: sil. \\\ na: eksik veri
#subset: sadece kritik satıra işlem yap.
#subset aynı zamanda 'hangi sütunlara göre karar verileceğini tam olarak belirtin.' anlamına gelir.
```

```python
numeric_cols = ["stars __count","forks __count","watchers __count","open __issues __count"]
for col in numeric_cols: #tüm sayısal sütunlarda herhangi bir değer stringse bunların tamamını sayısala çevir.
  if col in dt.columns: #yukarda yazılan gibi bir sayısal kolon var mı yok mu bunu kontrol eder.
    dt[col]=pd.to_numeric(dt[col],errors="coerce")
    #pd.to_numeric: bir sütunu sayıya çevirmeye çalışır.
    #errors="coerce":dönüştürülemeyen bir değer olursa eğer bunu NaN yapacak /// "abc" = NaN (bunu sayıya çeviremez.)
    #Peki neler yapabilir? "100"= 100 \\\ "50.5"= 50.5

    # Not:  NaN = Not a Number (sadece sayısal değerlerde kullanılır.) (numpy ile gelir. Kullanmak için numpy a ihtiyaç vardır.)
    #       NA = Not Available (bu veri yok.)
```
